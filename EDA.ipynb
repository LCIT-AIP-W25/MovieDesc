```python
# Merge all DataFrames into one
df_merged = pd.concat([df1, df2, df3, df4, df5, df6], ignore_index=True)

# Display merged DataFrame info
print(df_merged.info())
print(df_merged.head())
```

```python
# Get summary statistics
print(df_merged.describe())

# Check missing values
print(df_merged.isnull().sum())
```

```python
# Drop rows with missing values (if necessary)
df_merged = df_merged.dropna()

# Or fill missing values (if applicable)
df_merged.fillna("Unknown", inplace=True)  # For text columns
```

```python
#check for duplicate videoid entries
print(df_merged["videoid"].duplicated().sum())
```

```python
# remove duplicate videoid
df_merged = df_merged.drop_duplicates(subset="videoid", keep="first")
```

```python
import re

def duration_to_seconds(duration):
    match = re.match(r'PT(?:(\d+)H)?(?:(\d+)M)?(?:(\d+)S)?', duration)
    hours = int(match.group(1)) if match.group(1) else 0
    minutes = int(match.group(2)) if match.group(2) else 0
    seconds = int(match.group(3)) if match.group(3) else 0
    return hours * 3600 + minutes * 60 + seconds

df_merged["duration_seconds"] = df_merged["duration"].apply(duration_to_seconds)
print(df_merged[["duration", "duration_seconds"]].head())
```

```python
# Save processed data
df_merged.to_parquet("processed_movie_dataset.parquet", index=False)
```

METHODS FOR VIDEO DOWNLOAD

```python
import requests

def download_video(video_url, save_path):
    response = requests.get(video_url, stream=True)
    with open(save_path, "wb") as file:
        for chunk in response.iter_content(chunk_size=1024):
            if chunk:
                file.write(chunk)
    print(f"Downloaded: {save_path}")

video_url = df_merged["contentUrl"].iloc[0] + ".parquet"
video_url = "https://ak.picdn.net/shutterstock/videos/21179.parquet"  # Assuming .mp4 file
download_video(video_url, "video_sample.parquet")


```
